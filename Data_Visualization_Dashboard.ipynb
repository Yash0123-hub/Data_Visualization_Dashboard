```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
print("Libraries Imported Successfully!")
```

```python
import pandas as pd
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [15000, 18000, 22000, 20000, 25000, 30000],
    "Profit": [3000, 4000, 5000, 4500, 6000, 7000],
    "Region": ["North", "South", "East", "West", "North", "South"]
}
df = pd.DataFrame(data)
df
```

```python
fig = px.bar(
    df,
    x="Month",
    y="Sales",
    color="Region",
    title="Monthly Sales Dashboard"
)
fig.show()
```

```python
fig = px.line(
    df,
    x="Month",
    y="Profit",
    markers=True,
    title="Monthly Profit Trend"
)
fig.show()
```

```python
fig = px.pie(
    df,
    names="Region",
    values="Sales",
    title="Sales Distribution by Region"
)
fig.show()
```

```python
fig = px.scatter(
    df,
    x="Sales",
    y="Profit",
    color="Region",
    size="Sales",
    hover_name="Month",
    title="Sales vs Profit Dashboard"
)
fig.show()
```
